```julia
# Install and import the DataFrames package
using Pkg
Pkg.add("DataFrames")
using DataFrames

# Create a DataFrame
df = DataFrame(Name = ["Alice", "Bob", "Charlie"], Age = [25, 30, 35], Score = [85, 90, 95])

# Display the DataFrame
println(df)

# Basic data manipulation
df.Age .= df.Age .+ 1  # Increment age by 1
println("After updating age:")
println(df)

# Filter rows with a condition
filtered_df = filter(row -> row.Score > 85, df)
println("Filtered rows (Score > 85):")
println(filtered_df)


```

```julia
# Create an array
data = [1, 2, 3, 4, 5]

# Perform operations
squared = data .^ 2  # Element-wise square
println("Squared values:")
println(squared)

# Compute summary statistics
mean_value = mean(data)
println("Mean value:")
println(mean_value)
```

```julia
using Pkg
Pkg.add("RCall")
using RCall


# Create a Julia array
julia_array = [1, 2, 3, 4, 5]

# Pass it to R
@rput julia_array  # Sends `julia_array` to R as an R object

# Use the array in R
R"""
mean_value <- mean(julia_array)  # Compute the mean in R
print(mean_value)  # Print the result
"""

# Bring the R result back to Julia
@rget mean_value  # Brings `mean_value` from R to Julia
println("Mean value computed in R: ", mean_value)
```

# Use Python Libraries in Julia

## Example 1: Using NumPy in Julia
Here’s an example using the popular NumPy library for numerical computation.


```julia
using Pkg
Pkg.add("PyCall")
using PyCall

# Import the NumPy library
np = pyimport("numpy")

# Create a NumPy array
array = np.array([1, 2, 3, 4, 5])
println("NumPy array: ", array)

# Perform operations
squared = np.power(array, 2)  # Element-wise square
println("Squared values: ", squared)

# Compute the mean
mean_value = np.mean(array)
println("Mean value: ", mean_value)


```

## Example 2: Visualizing Data with Matplotlib

```julia
using PyCall

# Import Matplotlib
plt = pyimport("matplotlib.pyplot")

# Create data
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

# Plot the data
plt.plot(x, y, label="y = x^2", color="blue")
plt.title("Line Plot from Python's Matplotlib")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.legend()
plt.show()  # Display the plot
```

## Example 3: Sharing Data Between Julia and Python
You can seamlessly share data between Julia and Python. For example, you can pass Julia arrays to Python and vice versa.

Passing Julia Data to Python
+ Returning Python Data to Julia

```julia
julia_array = [1, 2, 3, 4, 5]  # Julia array

# Convert Julia array to Python list
py_array = PyObject(julia_array)

println("Julia array as Python object: ", py_array)


# ############

# Import NumPy
np = pyimport("numpy")

# Create a NumPy array
py_array = np.array([10, 20, 30, 40, 50])

# Convert Python array back to Julia
julia_array = collect(py_array)

println("Python array as Julia object: ", julia_array)
```

```julia
# Define a macro to generate mathematical functions
macro generate_math_function(op_symbol, func_name)
    quote
        # Dynamically define a new function
        function $(esc(func_name))(a, b)
            return a $(esc(op_symbol)) b
        end
    end
end

# Use the macro to create new functions
@generate_math_function :+ :'add'       # Creates a function `add(a, b) = a + b`
@generate_math_function :* :'multiply' # Creates a function `multiply(a, b) = a * b`

# Test the dynamically created functions
println(add(2, 3))        # Output: 5
println(multiply(4, 5))   # Output: 20
```
